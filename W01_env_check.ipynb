# W01 Environment Check

This notebook verifies the Week 1 machine-learning dependencies and reports the available compute device.

```python
import platform
import sys
print('Python executable:', sys.executable)
print('Python version:', sys.version.replace('\n', ' '))
print('Operating system:', platform.platform())
```

```python
from importlib import import_module
from importlib.metadata import PackageNotFoundError, version
packages = {'torch':'torch','stable_baselines3':'stable-baselines3','gymnasium':'gymnasium','librosa':'librosa','transformers':'transformers','sklearn':'scikit-learn','numpy':'numpy','matplotlib':'matplotlib'}
results=[]
for import_name, package_name in packages.items():
    try:
        import_module(import_name)
        try:
            package_version = version(package_name)
        except PackageNotFoundError:
            package_version = 'version unavailable'
        results.append((package_name,'OK',package_version))
    except Exception as exc:
        results.append((package_name,'FAILED',f'{type(exc).__name__}: {exc}'))
for package_name,status,details in results:
    print(f'{package_name:24} {status:7} {details}')
failed=[name for name,status,_ in results if status=='FAILED']
print('\nAll required core packages imported successfully.' if not failed else '\nMissing or broken packages: '+', '.join(failed))
```

```python
visual_backends=[]
for import_name, package_name in [('cv2','opencv-python'),('mediapipe','mediapipe')]:
    try:
        module=import_module(import_name)
        visual_backends.append((package_name,'OK',getattr(module,'__version__','version unavailable')))
    except Exception as exc:
        visual_backends.append((package_name,'NOT AVAILABLE',f'{type(exc).__name__}: {exc}'))
for package_name,status,details in visual_backends:
    print(f'{package_name:24} {status:13} {details}')
if not any(status=='OK' for _,status,_ in visual_backends):
    raise RuntimeError('Install at least one visual backend: opencv-python or mediapipe.')
```

```python
import torch
print('PyTorch version:', torch.__version__)
print('CUDA available:', torch.cuda.is_available())
if torch.cuda.is_available():
    print('CUDA device count:', torch.cuda.device_count())
    for index in range(torch.cuda.device_count()):
        print(f'CUDA device {index}:', torch.cuda.get_device_name(index))
    selected_device=torch.device('cuda')
elif getattr(torch.backends,'mps',None) and torch.backends.mps.is_available():
    print('Apple MPS available: True')
    selected_device=torch.device('mps')
else:
    print('No CUDA or Apple MPS accelerator detected; using CPU.')
    selected_device=torch.device('cpu')
print('Selected device:', selected_device)
```

```python
import gymnasium as gym
import numpy as np
from stable_baselines3 import PPO
env=gym.make('CartPole-v1')
observation,info=env.reset(seed=42)
action=env.action_space.sample()
next_observation,reward,terminated,truncated,info=env.step(action)
env.close()
print('Gymnasium smoke test: OK')
print('Stable Baselines3 PPO import: OK')
print('Observation shape:', np.asarray(observation).shape)
print('Sample reward:', reward)
```
